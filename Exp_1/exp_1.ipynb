```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("OpenCV Version:", cv2.__version__)
print("NumPy Version:", np.__version__)
```

```python
# Now Read the image 
img = cv2.imread("Dog.jpg")

if img is None:
  print("Image not found!")
else :
  print("Image loaded.....")
  print("Shape:",img.shape)
  print("Data Type:",img.dtype)
```

```python
# Now Display the image using OpenCV
cv2.imshow("Original_Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Now Display image using matplotlib

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,8))
plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("Off")
plt.show()
```

```python
# Now finding details of image
height, width, channels = img.shape

print("Height :", height)
print("Width :", width)
print("Channels:", channels)
```

```python
# Access Pixel
print("Pixel Value at (80,80):")
print(img[80,80])
```

```python
# Now we are Modifying the Pixel
modified = img.copy()
modified[80,80] = [0,255,0]

cv2.imshow("modified Pixel", modified)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Region of Interest (ROI/Crop)
roi = img [100:600,100:800]

cv2.imshow("ROI",roi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Now Converting the img to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Resize
resize = cv2.resize(img,(600,400))

cv2.imshow("Resized :", resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Flip the image
horizontal = cv2.flip(img,1)
vertical = cv2.flip(img,0)

cv2.imshow("Horizontal Flip",horizontal)
cv2.imshow("Vertical Flip", vertical)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Rotate the image
rotate = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)

cv2.imshow("Rotated :",rotate)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Draw Rectangle 
rectangle = img.copy()

cv2.rectangle(rectangle,
              (100,100),
              (400,300),
              (0,255,0),
              3)

cv2.imshow("Rectangle",rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Adding text 
text = img.copy()

cv2.putText(text,
            "Computer Vision Lab",
            (500,30),
            cv2.FONT_HERSHEY_SCRIPT_COMPLEX,
            1,
            (255,0,0),
            2)

cv2.imshow("Text :",text)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
