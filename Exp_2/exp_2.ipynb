```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("Dog.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap="gray")
```

```python
def add_salt_pepper(img , probability):
  noisy = img.copy()
  
  random = np.random.rand(*img.shape)
  
  noisy[random < probability/2]=0
  noisy[random > 1 - probability/2] = 255
  
  return noisy
```

```python
pb = 0.1
noisy_image = add_salt_pepper(image, pb)
plt.imshow(noisy_image, cmap="gray")
```
